```python
#membuat list dengan for
list = [i for i in range(0,20,2)]
print(list)
```

```python
#membuat list dengan if
list1 = [i for i in range(0,20,2) if i >= 2]
print("list 1:", list1)

list2 = [i for i in range(0,20) if i % 2 == 0]
print("list 2:", list2)

list3 = [i for i in range(0,20) if i % 2 != 0]
print("list 3:", list3)
```
