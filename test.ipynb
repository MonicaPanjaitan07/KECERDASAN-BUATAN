```python
def f(x,y):
    return np.cos(np.exp(-x**2-y**2))*np.sin(np.sqrt(x**2+y**2))

x = np.linspace(-np.pi, np.pi, 30)
y = np.linspace(-np.pi, np.pi, 30)

X, Y = np.meshgrid(x,y)
Z = f(X,Y)

fig = plt.figure()
ax = plt.axes(projection = '3d')
ax.contour3D(X, Y, Z, 50, cmap = 'binary')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z');
```

```python
import numpy as np
import matplotlib.pyplot as plt

# buat rentang sumbu x dan y
x_arr = np.linspace(-1.5, 1.5, 400)
y_arr = np.linspace(-1.5, 1.5, 400)
X, Y = np.meshgrid(x_arr, y_arr)

# f(x,y) = 0
F = X**6 + 3*X**4*Y**2 - 3*X**4 + 3*X**2*Y**4 - X**2*Y**3 - 6*X**2*Y**2 + 3*X**2 + Y**6 - 3*Y**4 + 3*Y**2 - 1

# plot kontur untuk level 0
plt.figure(figsize=(5,5))
plt.contour(X, Y, F, [0], colors='m', linewidths=2)

plt.title("Grafik 2D Persamaan Inplisit")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# tentukan grid koordinat
x_val = np.linspace(-2, 2, 100)
y_val = np.linspace(-2, 2, 100)
X, Y = np.meshgrid(x_val, y_val)

# hitung Z dari persamaan: ((5y/4) - |x|)^2 + z^2 = 1 (maka z = +- sqrt(1 - ...))
isi_akar = 1 - ((5 * Y / 4) - np.abs(X))**2

# hapus nilai negatif agar tidak error akar
isi_akar[isi_akar < 0] = np.nan
Z = np.sqrt(isi_akar)

# plot 3D
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# gambar permukaan atas dan bawah
ax.plot_surface(X, Y, Z, cmap='plasma')
ax.plot_surface(X, Y, -Z, cmap='plasma')

ax.set_title("Plot 3D")
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")

plt.show()
```
