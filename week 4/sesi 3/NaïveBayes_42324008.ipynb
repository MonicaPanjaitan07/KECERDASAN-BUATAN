```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris

# Load dataset Iris dari scikit-Learn
iris = load_iris()
```

### Pada langkah ini, saya mengimpor library numpy dan pandas yang digunakan untuk pengolahan data numerik dan tabel selanjutnya, saya memuat sataset Iris dari pustaka scikit-learn dan menyimpannya ke dalam variabel iris.


```python
# Konversi data menjadi DataFrame Pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]

# Definisikan X (fitur) dan y (target/Label)
X = df.drop(columns=["Species"])
y = df["Species"]
print(X)
```

### Di sini, saya mengonversi dataset mentah Iris menjadi bentuk DataFrame agar struktur datanya menjadi rapi dan mudah dibaca.

### Saya menambahkan kolom Species yang berisi nama-nama jenis bunga berdasarkan nilai targetnya.

### Kemudian, saya memisahkan dataset menjadi dua variabel utama, yaitu X untuk atribut/fitur ukuran bunga dan y untuk label/target spesies bunga.

```python
y
```

### Pada tahap ini, saya memanggil variabel y untuk menampilkan daftar label spesies yang ada pada dataset.

### Langkah ini bertujuan untuk memastikan bahwa data target telah terpisah dengan benar dari variabel X.

```python
from sklearn.preprocessing import LabelEncoder

lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

### Pada kode ini, saya menggunakan LabelEncoder untuk mengubah data label spesies yang awalnya berupa teks (string) menjadi format angka numerik (0, 1, 2).

### Perubahan ini dilakukan karena model pembelajaran mesin dapat memproses data dalam bentuk angka dengan lebih efisien.

```python
from sklearn.model_selection import train_test_split

X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

### Di sini, saya membagi dataset menjadi dua bagian menggunakan train_test_split, yaitu data latih (training) dan data uji (testing).

### Pembagian data dilakukan dengan proporsi 80% untuk pelatihan model (X_train, y_train) dan 20% untuk pengujian model (X_test, y_test).

```python
from sklearn.naive_bayes import GaussianNB

model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

### Pada tahap ini, saya mengimpor kelas GaussianNB dari pustaka sklearn.naive_bayes yang cocok digunakan untuk fitur bernilai numerik kontinu.   
### Saya membuat objek model klasifikasi dengan memanggil fungsi GaussianNB() dan menyimpannya ke dalam variabel model.   
### Selanjutnya, saya melatih model tersebut menggunakan metode .fit() dengan memasukkan data latih (X_train dan y_train) agar algoritma dapat mempelajari pola hubungan antara fitur dan labelnya.   
### Hasil proses pelatihan ini disimpan ke dalam variabel NB_model untuk digunakan pada tahap prediksi selanjutnya.

```python
result = NB_model.predict(X_test)
result
```

### Pada tahap ini, saya melakukan prediksi terhadap data uji (X_test) menggunakan model NB_model yang telah dilatih.   
### Fungsi .predict() akan mengembalikan hasil tebakan kelas berupa array angka (0, 1, atau 2) berdasarkan pola yang dipelajari.   
### Hasil prediksi tersebut kemudian disimpan ke dalam variabel result dan ditampilkan untuk diproses lebih lanjut. 

```python
from sklearn.metrics import accuracy_score

print('Accuracy of data testing: ', accuracy_score(y_test, result) * 100)
```

### Selanjutnya Saya menghitung tingkat akurasi model dengan membandingkan nilai label asli (y_test) terhadap hasil prediksi (result).   
### Nilai akurasi dikalikan 100 untuk menampilkan performa model dalam bentuk persentase, yang pada pengujian ini menghasilkan akurasi sebesar 96.67%.

```python
# prediksi inputan data baru
# input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
input_data = [[5.1, 3.5, 1.4, 0.2]]
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```
